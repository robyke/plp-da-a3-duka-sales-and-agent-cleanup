## Task 2: Create Derived Columns

I created the revenue, month, weekday, is_weekend and price_band columns to make the sales data easier to analyse.

```python
sales["revenue"] = sales["quantity"] * sales["unit_price"]

sales["month"] = sales["order_date"].dt.month_name()

sales["weekday"] = sales["order_date"].dt.day_name()

sales["is_weekend"] = sales["weekday"].isin(["Saturday", "Sunday"])

sales["price_band"] = np.where(
    sales["unit_price"] >= 28000,
    "Premium",
    "Value"
)

sales.head()
```

## Task 3: Branch Ranking by Revenue and Number of Orders

I compared each branch by total revenue and by number of orders to see whether the branch with the most orders also generated the most revenue.

```python
branch_summary = sales.groupby("branch").agg(
    total_revenue=("revenue", "sum"),
    orders=("order_id", "count")
).sort_values("total_revenue", ascending=False)

display(branch_summary)

print("\nRanking by total revenue:")
print(
    branch_summary["total_revenue"]
    .sort_values(ascending=False)
)

print("\nRanking by number of orders:")
print(
    branch_summary["orders"]
    .sort_values(ascending=False)
)
```

### Task 3 Answer

The branch rankings are the same for both total revenue and number of orders. Nairobi ranks first with 376 orders and KES 26,106,100 in revenue, followed by Mombasa with 186 orders and KES 14,003,700. Nakuru ranks third with 162 orders and KES 11,970,400, followed by Kisumu with 159 orders and KES 10,868,100, while Eldoret ranks last with 117 orders and KES 8,844,200. In this dataset, branches with more orders also generated more total revenue, although this relationship is not guaranteed because revenue also depends on the products purchased, quantities and unit prices.

## Task 4: Top Products by Revenue and Units Sold

I compared products by total revenue and by total units sold to determine whether the products generating the most revenue were also the products selling the most units.

```python
product_summary = sales.groupby("product").agg(
    total_revenue=("revenue", "sum"),
    units_sold=("quantity", "sum")
)

top_revenue = (
    product_summary["total_revenue"]
    .sort_values(ascending=False)
    .head(3)
)

top_units = (
    product_summary["units_sold"]
    .sort_values(ascending=False)
    .head(3)
)

print("Top 3 products by revenue:")
print(top_revenue)

print("\nTop 3 products by units sold:")
print(top_units)
```

### Task 4 Answer

The top three products by revenue are Laptop with KES 41,535,000, Tablet with KES 19,348,000 and Phone with KES 8,880,000. The top three products by units sold are Tablet with 691 units, Laptop with 639 units and Phone with 592 units. The lists contain the same three products, but the order is different because revenue depends on both the number of units sold and the unit price. Laptops generate the highest revenue even though Tablets sold more units because Laptop has a much higher unit price.

## Task 5: Weekend vs Weekday Orders

I compared weekend and weekday sales using the number of orders, total revenue and average order revenue.

```python
weekend_summary = sales.groupby("is_weekend").agg(
    orders=("order_id", "count"),
    total_revenue=("revenue", "sum"),
    average_order_revenue=("revenue", "mean")
)

weekend_summary.index = weekend_summary.index.map({
    False: "Weekday",
    True: "Weekend"
})

display(weekend_summary.round(2))
```

### Task 5 Answer

Weekday orders were much more common than weekend orders. There were 723 weekday orders generating KES 51,856,200 in total revenue, compared with 277 weekend orders generating KES 19,936,300. However, the average order revenue was very similar: KES 71,723.65 on weekdays and KES 71,972.20 on weekends. This means weekdays generated more total revenue mainly because there were more orders, not because each order was worth more.

## Task 6: Best and Worst Revenue Month

I compared monthly revenue to identify the highest-performing and lowest-performing months and calculate the percentage difference between them.

```python
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June"
]

monthly_revenue = (
    sales.groupby("month")["revenue"]
    .sum()
    .reindex(month_order)
)

print(monthly_revenue)

best_month = monthly_revenue.idxmax()
best_revenue = monthly_revenue.max()

worst_month = monthly_revenue.idxmin()
worst_revenue = monthly_revenue.min()

percentage_difference = (
    (best_revenue - worst_revenue) / worst_revenue * 100
)

print("\nBest month:", best_month)
print("Best month revenue:", best_revenue)

print("\nWorst month:", worst_month)
print("Worst month revenue:", worst_revenue)

print(
    "\nPercentage difference:",
    round(percentage_difference, 1),
    "%"
)
```

### Task 6 Answer

April was the best-performing month with total revenue of KES 13,518,300. June was the lowest-performing month with KES 10,903,400. April's revenue was approximately 24.0% higher than June's revenue. This shows that monthly performance varied meaningfully across the first half of 2026, with April producing the strongest sales result and June the weakest.

## Task 7: Premium Online Orders from Nairobi and Mombasa

I filtered the data to keep only Premium products sold through the Online channel in Nairobi and Mombasa, then calculated the number of orders and total revenue and exported the result.

```python
premium_online = sales[
    (sales["price_band"] == "Premium")
    & (sales["channel"] == "Online")
    & (sales["branch"].isin(["Nairobi", "Mombasa"]))
].copy()

premium_count = len(premium_online)
premium_revenue = premium_online["revenue"].sum()

print("Number of Premium online orders:", premium_count)
print("Total revenue:", premium_revenue)

premium_online.to_csv(
    "outputs/premium_online_coast_capital.csv",
    index=False
)

display(premium_online.head())
```

### Task 7 Answer

There were 83 Premium online orders from Nairobi and Mombasa. Together, these orders generated KES 10,218,000 in total revenue. The filtered results were exported to `outputs/premium_online_coast_capital.csv` with the index excluded, as required.

```python
from pathlib import Path

# Create excel folder if it does not exist
Path("excel").mkdir(exist_ok=True)

# Region mapping used for the XLOOKUP equivalent
region_map = {
    "Nairobi": "Central",
    "Mombasa": "Coast",
    "Kisumu": "Western",
    "Nakuru": "Rift Valley",
    "Eldoret": "North Rift",
}

excel_sales = sales.copy()
excel_sales["region"] = excel_sales["branch"].map(region_map)

# Branch revenue summary
branch_excel = (
    excel_sales.groupby("branch", as_index=False)
    .agg(total_revenue=("revenue", "sum"))
    .sort_values("total_revenue", ascending=False)
)

# Revenue by branch and channel
pivot_excel = pd.pivot_table(
    excel_sales,
    index="branch",
    columns="channel",
    values="revenue",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name="Total"
)

# Save workbook
with pd.ExcelWriter(
    "excel/duka_branch_summary.xlsx",
    engine="openpyxl"
) as writer:
    excel_sales.to_excel(
        writer,
        sheet_name="Sales Data",
        index=False
    )

    branch_excel.to_excel(
        writer,
        sheet_name="Branch Summary",
        index=False
    )

    pivot_excel.to_excel(
        writer,
        sheet_name="Branch Channel Pivot"
    )

print("Created excel/duka_branch_summary.xlsx")

print("\nBranch Summary:")
display(branch_excel)

print("\nBranch x Channel Revenue Pivot:")
display(pivot_excel)
```

### Task 8: Spreadsheet Validation

The Excel workbook was saved as `excel/duka_branch_summary.xlsx`. It contains the sales data with a region column, a branch revenue summary, and a revenue-by-branch-and-channel pivot table. The branch revenue totals in the workbook match the pandas results from the notebook.
